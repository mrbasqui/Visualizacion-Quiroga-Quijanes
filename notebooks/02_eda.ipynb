# 02 - Análisis Exploratorio de Datos (EDA)
Este notebook aborda el análisis univariado, relaciones entre variables y componente temporal, preparando los hallazgos que se visualizan en Streamlit.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/processed/accidentes_procesados_2019_2023.csv')
df['Fecha_Hora'] = pd.to_datetime(df['Fecha_Hora'])
```

### 2. Análisis Univariado

```python
# Frecuencia de accidentes por región
plt.figure(figsize=(10,6))
df['Region'].value_counts().sort_values().plot(kind='barh', color='salmon')
plt.title('Total de Siniestros por Región')
plt.xlabel('Cantidad')
plt.show()
```

### 3. Relaciones entre variables (X vs Y)

```python
# Relación entre Zona y Víctimas Fatales
df.groupby('Zona')['Fallecidos'].mean().plot(kind='bar', color=['skyblue', 'lightgreen'])
plt.title('Promedio de Fallecidos por Siniestro según Zona')
plt.ylabel('Promedio de Fallecidos')
plt.show()
```

### 4. Componente Temporal (T)

```python
# Evolución mensual de siniestros
df['Año_Mes'] = df['Fecha_Hora'].dt.to_period('M')
df.groupby('Año_Mes')['Total_Siniestros'].count().plot(kind='line', marker='o', figsize=(12,5))
plt.title('Evolución Mensual de Accidentes de Tránsito')
plt.ylabel('Cantidad de Siniestros')
plt.grid(True)
plt.show()
```
